```python
# 1. 데이터 불러오기
import pandas as pd
url='https://raw.githubusercontent.com/sehakflower/data/main/titanic.csv'
titanic_df= pd.read_csv(url,sep='\t')
titanic_df.info()
```

```python
# 컬럼명 변경
new_columns=['passengerId', 'survived', 'pclass', 'name', 'sex', 'age', 'sibsp', 'parch', 'ticket', 'fare', 'cabin', 'embarked']
titanic_df.columns=new_columns
titanic_df.columns
```

```python
# isna 함수
titanic_df['age'][titanic_df['age'].isna()]
```

```python
# isnull 함수
titanic_df['age'].isnull()
```

```python
# inull 함수로 결측치인 NaN 값 추출
titanic_df['age'][titanic_df['age'].isnull()]
```

```python
# NaN 값 개수 확인
titanic_df['age'][titanic_df['age'].isnull()].isnull().sum()
```

```python
# notnull 함수
titanic_df['age'].notnull()
```

```python
# notnull로 NaN이 아닌값만 추출
titanic_df['age'][titanic_df['age'].notnull()]
```

```python
# notnull 활용해 age 열에서 NaN이 아닌 데이터를 검색, age와 name 출력
titanic_df.loc[titanic_df['age'].notnull(),['age', 'name']]
```

```python
# notnull 활용해 결측치가 있는 행을 추출하고 age열에서 결측치인 행과 해당하는 name 열의 값을 추출
titanic_df.loc[titanic_df['age'].isnull(),['age', 'name']]
```

```python
# 결측치 처리 
titanic_df.head()
```

```python
# cabin 열의 결측치를 'N'으로 변경
titanic_df['cabin']=titanic_df['cabin'].fillna('N')
titanic_df['cabin']
```

```python
# cabin 열 값의 개수 확인
titanic_df['cabin'].value_counts()
```

```python
# 첫 번째 인덱스 값 확인
titanic_df['cabin'][0]
```

```python
# apply와 lambda 식을 이용해 데이터의 첫 문자만 남기기 
titanic_df['cabin']=titanic_df['cabin'].apply(lambda x : x[0])
titanic_df['cabin']
```

```python
titanic_df.info()
```

```python
# 데이터프레임(titanic_df) 복제
titanic_df_copy=titanic_df.copy()
titanic_df_copy
```

```python
# 결측치를 찾는(True로 표시) isnull 함수를 이용
titanic_df_copy['age'].isnull()
```

```python
# mean 함수로 열의 평균값 구하고 fillna 함수로 결측치 채우기
titanic_df_copy['age']=titanic_df_copy['age'].fillna(titanic_df_copy['age'].mean())
titanic_df_copy['age']
```

```python
# copy(원본 데이터프레임을 유지한 채 새로운 주소로 복사)
copy_titanic_df=titanic_df.copy()
copy_titanic_df
```

```python
# 위 코드와 동일함
hex(id(copy_titanic_df))
```
