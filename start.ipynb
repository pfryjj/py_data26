```python
number_list = [23, 45, 27, 11, 25, 65, 78]

def getIndex(num_list, target):
    for a in range(len(num_list)):
        if num_list[a] == target:
            return a

def getMax(num_list):
    result = num_list[0]
    for num in num_list:
        if num > result:
            result = num
    return result

def getMin(num_list):
    result = num_list[0]
    for num in num_list:
        if num < result:
            result = num
    return result

def countGT(num_list, target):
    count = 0
    for num in num_list:
        if num > target:
            count += 1
    return count

def sumList(num_list):
    s = 0
    for num in num_list:
        s += num
    return s

def swapList(num_list):
    for a in range(len(num_list)//2):
        num_list[a], num_list[len(num_list)-1-a] = num_list[len(num_list)-1-a], num_list[a]

print(getIndex(number_list, 25))
print(getMax(number_list))
print(getMin(number_list))
print(countGT(number_list, 42))
print(sumList(number_list))
swapList(number_list)
print(number_list)
```
